
### 2 Viewing & Inspecting Data


```python
import pandas as pd
import numpy as np

# Fallback definition since 'file.csv' was missing and 'df' was not created
try:
    df
except NameError:
    df = pd.DataFrame(np.random.rand(5, 4), columns=['A', 'B', 'C', 'D'])

print(df.head(5))                          # First 5 rows
print(df.tail(5))                          # Last 5 rows
df.info()                                  # Index, datatypes, memory info, non-null counts
print(df.describe())                       # Summary statistics for numerical columns
print(df.shape)                            # Return tuple of (rows, columns)
```

### 3 Selection & Filtering

```python
df['A']                             # Select column 'A' (as Series)
df[['A', 'B']]                      # Select columns 'A' and 'B' (as DataFrame)
df.iloc[0:5, 0:2]                   # Select rows 0-4 and columns 0-1 by index position
df.loc[df['A'] > 0.5, ['A', 'B']]   # Filter rows by condition and select columns by label
```

### 4 Cleaning & Transformation

```python
df.dropna()                         # Drop all rows containing null values
df.fillna(value=0)                  # Replace missing values with 0
df.drop(['A'], axis=1)              # Drop column 'A'
df.rename(columns={'B': 'Beta'})    # Rename column 'B' to 'Beta'
df.drop_duplicates()                # Remove duplicate rows
df['A'].apply(lambda x: x * 2)      # Apply custom function to column 'A'
```

### 5 Aggregation & Joins

```python
import pandas as pd
import numpy as np

# Ensure df has a 'Category' column for demonstration
if 'Category' not in df.columns:
    df['Category'] = ['X', 'Y', 'X', 'Y', 'Z']

# Define dummy df1 and df2 for the merge and concat demonstrations
df1 = pd.DataFrame({'key_col': ['foo', 'bar'], 'val1': [1, 2]})
df2 = pd.DataFrame({'key_col': ['foo', 'baz'], 'val2': [3, 4]})

# Run demonstrations
print("--- Groupby Result ---")
print(df.groupby('Category').mean(numeric_only=True))       # Group by column and compute numerical mean

print("\n--- Pivot Table Result ---")
print(df.pivot_table(index='A', values='B', aggfunc='median')) # Generate pivot table

print("\n--- Merge Result ---")
print(pd.merge(df1, df2, on='key_col'))    # SQL-style merge/join on a matching key column

print("\n--- Concat Result ---")
print(pd.concat([df1, df2], axis=0))       # Append rows of df2 to df1
```
