# Matplotlib & Seaborn : Data Visualization
Matplotlib provides low-level, highly customizable plots, while Seaborn simplifies statistical visualizations.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

# 1 Line Plot

```python
# 1. Line Plot

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
y2 = [12, 18, 22, 20, 35]

plt.plot(x, y)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Define variables x and y
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.figure()
plt.scatter(x, y, color='blue', marker='o')
plt.title('Scatter Plot of X vs Y')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.grid(True)
plt.show()
```

```python
plt.figure()
plt.plot(x, y, color='red', marker='o', label='Data 1')
plt.plot(x, y2, linestyle="--", label="Data 2")
plt.title("Simple Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.legend()
plt.show()
```

# 2 Multiple Lines

```python
plt.figure()
plt.plot(x,y,label="2x")
plt.plot(x,y2,label="x²")
plt.plot(x,y2,label="Data 2")
plt.legend()
plt.grid(True)
plt.show()
```

# 3 Bar Chart

```python
plt.figure()
plt.bar(x,y,color="orange")
plt.title("Bar Chart")
plt.show()
```

```python
# Horizontal Bar Chart
plt.figure()
plt.barh(x,y,color="purple")
plt.show()
```

```python
data=np.random.randn(1000)
plt.figure()
plt.hist(data,bins=20,color="skyblue")
plt.title("Histogram")
plt.show()
```

```python
labels=["Python","Java","C++","AI"]
sizes=[40,20,15,25]
plt.figure()
plt.pie(sizes,labels=labels,autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show()
```

```python
plt.figure()
plt.violinplot(data)
plt.title("Violin Plot")
plt.show()
```

```python
plt.figure()
plt.fill_between(x,y,color="lightblue")
plt.title("Area Plot")
plt.show()
```

```python
# 30. Pandas Plot
df=pd.DataFrame({
    "Month":["Jan","Feb","Mar","Apr"],
    "Sales":[100,120,150,180]
})

df.plot(x="Month",y="Sales",kind="line")
plt.show()
```

```python
# 31. Real Dataset Example
sales=pd.DataFrame({
    "Month":["Jan","Feb","Mar","Apr","May"],
    "Profit":[20,30,45,35,55]
})
plt.figure()
plt.plot(sales["Month"],sales["Profit"],marker="o")
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.grid(True)
plt.show()
```
